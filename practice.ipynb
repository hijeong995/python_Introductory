## 리스트 변수
1. 리스트 선언
```
    - 리스트변수 = []
    - 리스트변수 = list()
    - 리스트변수 = [데이터1, 데이터2, ...]
```
2. 리스트 추가
```
    - 리스트변수.append(데이터)
    - 리스트변수.insert(인덱스번호, 데이터)
```
3. 리스트 삭제
```
    - 리스트변수.remove(데이터)
    - del 리스트변수[인덱스번호]
```
4. 리스트 데이터 수정
```
    - 리스트변수[인덱스번호] = 수정할 데이터
```
